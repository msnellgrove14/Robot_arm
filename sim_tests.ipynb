## Imports

```python
import pinocchio as pin  ##type: ignore
from pinocchio.visualize import MeshcatVisualizer  ##type: ignore
import numpy as np
import sys
import time

from trajectories import (
    linear_trajectory,
    circular_trajectory,
    figure_eight_trajectory,
    helical_trajectory,
    oscillating_orientation_trajectory,
    helical_with_wobble_trajectory,
    lissajous_3d_trajectory,
    tumbling_lissajous_trajectory,
    chaotic_fourier_trajectory,
    orbital_trajectory,
)
from simulators import simulate_pd, simulate_lqr
```

## Load model

```python
model, collision_model, visual_model = pin.buildModelsFromUrdf(
    "lbr_iiwa7_r800.urdf", package_dirs="."
)
data = model.createData()
```

## Initialize visualizer (meshcat)

```python
try:
    viz = MeshcatVisualizer(model, collision_model, visual_model)
    viz.initViewer(open=True)
    viz.loadViewerModel()
except ImportError as err:
    print(
        "Error while initializing the viewer. "
        "It seems you should install Python meshcat"
    )
    print(err)
    sys.exit(0)
```

## 1. Linear trajectory

```python
T, dt = 5.0, 0.001
poses = linear_trajectory(T=T, dt=dt)
```

### PD

```python
pd_res = simulate_pd(model, poses, T, dt, viz=viz, viz_every=10, real_time=True)
print(f"max error: {pd_res['errors'].max():.4f}  IK converged: {sum(pd_res['successes'])}/{len(pd_res['successes'])}")
```

### LQR

```python
lqr_res = simulate_lqr(model, poses, T, dt, viz=viz, viz_every=10, real_time=True)
print(f"max error: {lqr_res['errors'].max():.4f}  IK converged: {sum(lqr_res['successes'])}/{len(lqr_res['successes'])}")
```

## 10. Orbital trajectory (orbits the robot's own base axis, radius+height breathe at different frequencies, orientation tangent to the path)

NOTE: joint1's real limit is +-170 deg, less than a full revolution -- this is a kinematic-only demo (not achievable on real hardware). Also note the tangent-orientation construction needs a continuity-preserving frame (see `orbital_trajectory`'s docstring) -- an earlier naive version had a real discontinuity that blew up the simulation; fixed and verified before this block was added.

```python
T, dt = 6.0, 0.001
poses = orbital_trajectory(
    r0=0.5, r_amplitude=0.15, r_frequency=2,
    z0=0.5, z_amplitude=0.12, z_frequency=3,
    T=T, dt=dt, n_orbits=1.0,
)
```

### PD

```python
pd_res = simulate_pd(model, poses, T, dt, viz=viz, viz_every=10, real_time=True)
print(f"max error: {pd_res['errors'].max():.4f}  IK converged: {sum(pd_res['successes'])}/{len(pd_res['successes'])}")
```
